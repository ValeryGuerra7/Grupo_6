# CREACIÓN DE LA CLASE VALIDADOR



```python
# ============================================================
#  BLOQUE 1 - LIBRERIAS, CONSTANTES Y VALIDACIONES GENERALES
# ============================================================
# Solo se usan librerias estandar de Python, por lo que el
# proyecto se ejecuta sin instalar nada adicional.

from datetime import datetime, date, timedelta

# --- Constantes de configuracion de la clinica -------------
FORMATO_FECHA = "%d/%m/%Y"   # formato con el que el usuario escribe las fechas
FORMATO_HORA = "%H:%M"       # formato con el que el usuario escribe las horas
HORA_APERTURA = 8            # la clinica abre 08:00
HORA_CIERRE = 18             # la clinica cierra 18:00


class Validador:
    """Agrupa las validaciones que se repiten en todo el sistema.

    Se usan metodos estaticos (@staticmethod) porque estas
    operaciones no necesitan los datos de un objeto en concreto:
    reciben un valor, lo revisan y lo devuelven limpio.
    Si el dato es incorrecto se lanza un ValueError para que el
    programa no siga trabajando con informacion invalida.
    """

    @staticmethod
    def texto(valor, campo, minimo=2):
        """Verifica que un texto no este vacio y lo devuelve capitalizado."""
        if not isinstance(valor, str) or len(valor.strip()) < minimo:
            raise ValueError(f"El campo '{campo}' debe tener al menos {minimo} caracteres.")
        return valor.strip().title()

    @staticmethod
    def frase(valor, campo, minimo=4):
        """Valida textos largos (motivo, diagnostico, tratamiento).

        A diferencia de texto(), aqui solo se pone en mayuscula la
        primera letra para no deformar palabras como '500mg'.
        """
        if not isinstance(valor, str) or len(valor.strip()) < minimo:
            raise ValueError(f"El campo '{campo}' debe tener al menos {minimo} caracteres.")
        limpio = valor.strip()
        return limpio[0].upper() + limpio[1:]

    @staticmethod
    def dni(valor):
        """El DNI peruano tiene 8 digitos numericos."""
        valor = str(valor).strip()
        if not valor.isdigit() or len(valor) != 8:
            raise ValueError("El DNI debe tener exactamente 8 digitos numericos.")
        return valor

    @staticmethod
    def telefono(valor):
        """El celular tiene 9 digitos."""
        valor = str(valor).strip()
        if not valor.isdigit() or len(valor) != 9:
            raise ValueError("El telefono debe tener 9 digitos numericos.")
        return valor

    @staticmethod
    def monto(valor, campo):
        """Los montos de dinero no pueden ser negativos ni cero."""
        try:
            valor = float(valor)
        except (TypeError, ValueError):
            raise ValueError(f"El campo '{campo}' debe ser un numero.")
        if valor <= 0:
            raise ValueError(f"El campo '{campo}' debe ser mayor a cero.")
        return round(valor, 2)

    @staticmethod
    def fecha(texto):
        """Convierte un texto 'dd/mm/aaaa' en un objeto date."""
        try:
            return datetime.strptime(str(texto).strip(), FORMATO_FECHA).date()
        except ValueError:
            raise ValueError(f"Fecha invalida '{texto}'. Use el formato dd/mm/aaaa.")

    @staticmethod
    def hora(texto):
        """Convierte un texto 'HH:MM' en un objeto time y revisa el horario."""
        try:
            momento = datetime.strptime(str(texto).strip(), FORMATO_HORA).time()
        except ValueError:
            raise ValueError(f"Hora invalida '{texto}'. Use el formato HH:MM.")
        if not (HORA_APERTURA <= momento.hour < HORA_CIERRE):
            raise ValueError(
                f"Fuera de horario. La clinica atiende de {HORA_APERTURA:02d}:00 a {HORA_CIERRE:02d}:00."
            )
        if momento.minute not in (0, 30):
            raise ValueError("Las citas se programan cada 30 minutos (ejemplo: 09:00 o 09:30).")
        return momento


def titulo(texto):
    """Imprime un titulo con separadores para que la salida se lea ordenada."""
    print("\n" + "=" * 62)
    print(texto.upper().center(62))
    print("=" * 62)


print("Bloque 1 cargado: constantes y clase Validador listas.")
```
